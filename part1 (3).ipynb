## 4: Filter DataFrame

```python
df = df[['continent', 'location', 'date', 'total_cases', 
         'total_deaths', 'gdp_per_capita', 'human_development_index']]
df.head()
```

## 5: Data Cleaning
### a. Remove duplicates

```python
df.drop_duplicates(inplace=True)
```

### b. Find missing values

```python
df.isnull().sum()
```

### c. Remove rows where continent is missing

```python
df.dropna(subset=['continent'], inplace=True)
```

### d. Fill remaining missing values with 0

```python
df.fillna(0, inplace=True)
```

## 6: Date Time Format
### a. Convert date column to datetime

```python
df['date'] = pd.to_datetime(df['date'], format='%d/%m/%Y')
df['date'].head()
```

### b. Create new column 'month' by extracting month

```python
df['month'] = df['date'].dt.month
df.head()
```

## 7: Data Aggregation
### a. Group by continent and find max value

```python
df_groupby = df.groupby('continent').max().reset_index()

# Display the new DataFrame
df_groupby.head()
```

## 8: Feature Engineering

```python
# Create a new feature 'total_deaths_to_total_cases'
df_groupby['total_deaths_to_total_cases'] = (
    df_groupby['total_deaths'] / df_groupby['total_cases']
)
df_groupby.head()
```

## 9: Data Visualization
### a. Histogram of gdp_per_capita

```python
sns.displot(df_groupby['gdp_per_capita'])
plt.show()
```

### b. Scatter plot of total_cases vs gdp_per_capita

```python
sns.scatterplot(x='total_cases', y='gdp_per_capita', data=df_groupby)
plt.show()
```

### c. Pairplot

```python
sns.pairplot(df_groupby)
plt.show()
```

### d. Bar plot of continent vs total_cases

```python
sns.catplot(x='continent', y='total_cases', kind='bar', data=df_groupby)
plt.show()
```

## 10: Save the DataFrame

```python
df_groupby.to_csv('df_groupby.csv', index=False)
pd.read_csv('df_groupby.csv').head()
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/SR1608/Datasets/main/covid-data.csv"
df = pd.read_csv(url)

# Keep useful columns
df = df[['continent', 'location', 'date', 'total_cases', 
         'total_deaths', 'gdp_per_capita', 'human_development_index']]

# Clean data
df.drop_duplicates(inplace=True)
df.dropna(subset=['continent'], inplace=True)
df.fillna(0, inplace=True)

df['date'] = pd.to_datetime(df['date'], format='%d/%m/%y')

# Extract month
df['month'] = df['date'].dt.month

print("Data cleaned and date converted successfully!")
```

```python
df.to_csv('covid_cleaned.csv', index=False)
print("Cleaned dataset saved as 'covid_cleaned.csv'")
```
